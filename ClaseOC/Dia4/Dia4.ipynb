## 1. Librerías (Herramientas listas)

Python tiene miles de herramientas ya creadas. No reinventemos la rueda. Para usarlas, usamos `import`.

```python
import time  # Librería para manejar el tiempo
import math  # Librería para matemáticas complejas

print("Iniciando cuenta regresiva...")
time.sleep(1) # Pausa de 1 segundo
print("3...")
time.sleep(1)
print("2...")
time.sleep(1)
print("1...")

print("La raíz cuadrada de 49 es:", math.sqrt(49))
```

## 2. Manejo de Errores (`try` / `except`)

¿Qué pasa si le pides un número al usuario y escribe "hola"? El programa explota (Crash). Para evitarlo, usamos bloques de protección.

```python
try:
    # Intentamos hacer algo peligroso
    numero = int(input("Por favor, ingresa un número para dividir 100: "))
    resultado = 100 / numero
    print("El resultado es:", resultado)
    
except ValueError:
    # Esto se ejecuta si el usuario escribe texto
    print("❌ Error: ¡Debes escribir un NÚMERO, no letras!")
    
except ZeroDivisionError:
    # Esto se ejecuta si escribe 0
    print("❌ Error: No se puede dividir por cero.")
    
print("El programa continúa sin explotar.")
```

## 🏆 Reto Semanal: El Cajero Automático

Vamos a simular un ATM usando funciones y manejo de errores.

**Consigna:**
1. Define una variable global `saldo = 1000`.
2. Crea una función `verificar_pin(pin_ingresado)` que devuelva `True` si el pin es "1234" y `False` si no.
3. Crea una función `retirar(cantidad)` que:
    * Reste la cantidad al saldo SOLO SI hay suficiente dinero.
    * Devuelva un mensaje de éxito o de "Fondos insuficientes".
4. En el bloque principal:
    * Pide el PIN.
    * Si es correcto, pide cuánto retirar (Usa `try/except` por si escribe letras).
    * Muestra el saldo final.

```python
saldo = 1000
l = True
while l == True:
    def verificar_pin(pin_ingresado):
        pin_correcto = 1234
        if pin_ingresado == pin_correcto:
            return True
        else:
            return False
    def retirar(cantidad):
        global saldo
        if cantidad <= saldo:
            saldo = saldo-cantidad
            print(f"Has retirado {cantidad}. Saldo restante: {saldo}")
        else:
            print("❌ Error: Fondos insuficientes.")
    print("Bienvenido al cajero automático.")
    try:
        pin_ingresado = int(input("Por favor, ingresa tu PIN: "))
        if verificar_pin(pin_ingresado):
            print(f"Tienes ${saldo} de saldo")
            cantidad = int(input("¿Cuánto deseas retirar? "))
            retirar(cantidad)
        else:
            print("❌ Error: PIN incorrecto.")
    except ValueError:
        print("❌ Error: Debes ingresar un número válido para la cantidad a retirar.")
    while True:
        y_n = input("Listo? (Y/N): ")
        if y_n == "Y":
            l = False
            break
        elif y_n == "N":
            break
        else:
            print("Opción inválida. Por favor, ingrese 'Y' o 'N'.")
```
