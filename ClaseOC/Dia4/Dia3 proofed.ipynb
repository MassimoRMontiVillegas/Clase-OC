## 1. Funciones (`def`)

En Python, eso es una función `def`.

```python
# Definimos la función (La enseñamos a Python)
def saludar(nombre):
    print("Hola", nombre)
    print("¡Espero que tengas un buen día!")
nombre=input("como te llamas :")
saludar(nombre)
```

### 1.1 `return` vs `print`

Esto es vital:
* `print`: Muestra algo en pantalla, pero el dato se pierde en el aire.
* `return`: La función te **devuelve** el dato para que lo guardes en una variable y lo sigas usando.

```python
# Función que SOLO muestra (print)
def sumar_pantalla(a, b):
    print(a + b)

# Función que DEVUELVE el valor (return)
def sumar_real(a, b):
    resultado = a + b
    return resultado


#Solución a error en el código del profe para ej.
valor=sumar_real(24,6)
print(valor+10)
```

### 🌡️ Ejercicio: Convertidor de Temperatura

Crea una función llamada `celsius_a_fahrenheit`.
1. Debe recibir los grados Celsius como parámetro.
2. Debe aplicar la fórmula: $(C \times 9/5) + 32$.
3. Debe hacer `return` del resultado.

```python
# Define tu función aquí
def celsius_a_fahrenheit(celsius):
    fahrenheit = (celsius * 9/5) + 32
    return fahrenheit
l = True
while l == True:
    try:
        celsius = float(input("Ingresa la temperatura en grados Celsius: "))
        fahrenheit = celsius_a_fahrenheit(celsius)
        print(f"{celsius}°C son {fahrenheit}°F")
    except ValueError:
        print("Dame números, no letras")
    while True:
        y_n = input("Listo? (Y/N): ")
        if y_n == "Y":
            l = False
            break
        elif y_n == "N":
            break
        else:
            print("Opción inválida. Por favor, ingrese 'Y' o 'N'.")
```

### 4. Mayor de Tres

**Consigna:**
1. Crea una función `mayor_de_tres(n1, n2, n3)`.
2. Usa condicionales (`if/elif/else`) para ver cuál es el más grande.
3. Retorna el número mayor.

```python
# Tu código aquí 👇
def mayorDeTres(n1, n2, n3):
    if n1 >= n2 and n1 >= n3:
        return n1
    elif n2 >= n1 and n2 >= n3:
        return n2
    else:
        return n3
l = True
while l == True:
    try:
        n1 = int(input("Ingresa el primer número: "))
        n2 = int(input("Ingresa el segundo número: "))
        n3 = int(input("Ingresa el tercer número: "))
        print(mayorDeTres(n1, n2, n3))
    except ValueError:
        print("Dame números, no letras.")
    while True:
        y_n = input("Listo? (Y/N): ")
        if y_n == "Y":
            l = False
            break
        elif y_n == "N":
            break
        else:
            print("Opción inválida. Por favor, ingrese 'Y' o 'N'.")
```

### 6. Conversor de Minutos a Horas

**Consigna:**
1. Crea una función `minutos_a_horas(minutos)`.
2. Calcula cuántas horas y cuántos minutos sobran (usa división entera `//` y módulo `%`).
3. Imprime algo como: **"2 horas y 30 minutos"**.

```python
# Tu código aquí 👇
def minutosAHoras(minutos):
    horas = minutos // 60
    minutos_restantes = minutos % 60
    return horas, minutos_restantes
l = True
while l == True:
    try:
        minutos = int(input("Ingresa la cantidad de minutos: "))
        horas, minutos_restantes = minutosAHoras(minutos)
        print(f"{minutos} minutos son {horas} horas y {minutos_restantes} minutos")
    except ValueError:
        print("Dame números, no letras.")
    while True:
        y_n = input("Listo? (Y/N): ")
        if y_n == "Y":
            l = False
            break
        elif y_n == "N":
            break
        else:
            print("Opción inválida. Por favor, ingrese 'Y' o 'N'.")
```

### 8. Calculadora de Descuento

**Consigna:**
1. Crea una función `calcular_precio_final(precio, descuento)`.
2. Calcula el precio con el descuento aplicado.
3. Retorna el precio final.

```python
# Tu código aquí 👇
def calcularPrecioFinal(precio, descuento):
    precio_final = precio - (precio * (descuento / 100))
    return precio_final
l = True
while l == True:
    try:
        precio = float(input("Ingresa el precio del producto: "))
        descuento = float(input("Ingresa el porcentaje de descuento: "))
        print(f"El precio final después del descuento es: {calcularPrecioFinal(precio, descuento)}")
    except ValueError:
        print("Dame números, no letras.")
    while True:
        y_n = input("Listo? (Y/N): ")
        if y_n == "Y":
            l = False
            break
        elif y_n == "N":
            break
        else:
            print("Opción inválida. Por favor, ingrese 'Y' o 'N'.")
```
